# Falcon 9 landing success through time

This authored analysis asks how landing outcomes changed as the program matured, and whether payload, orbit, reuse, and launch site are separable from historical era. The data is validated and queried through the packaged DuckDB analytics layer.

```python
import plotly.express as px

from spacex_falcon.analytics import modeling_frame, site_success, yearly_success
from spacex_falcon.data import load_launch_data

launches = load_launch_data()
analysis = modeling_frame(launches)
analysis.shape
```

## Program maturation

Landing success rises strongly through time. This makes random train and test splits unsuitable because they mix early and mature program eras.

```python
annual = yearly_success(launches)
px.line(annual, x='LaunchYear', y='SuccessRate', markers=True, range_y=[0, 1])
```

## Site and reuse patterns

Launch sites entered service in different eras, so site level success rates combine location effects with temporal change. Booster reuse and hardware configuration show the same confounding risk.

```python
display(site_success(launches))
px.scatter(analysis, x='FlightNumber', y='PayloadMass', color='Class', symbol='Reused', hover_data=['Orbit', 'LaunchSite', 'Flights'])
```
