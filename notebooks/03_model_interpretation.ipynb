# Calibration, interpretation, and error analysis

A probability is useful only when its uncertainty and failure modes are visible. This notebook compares calibration, inspects feature contributions, and reviews incorrect out of period predictions.

```python
from spacex_falcon.analytics import modeling_frame
from spacex_falcon.data import load_launch_data
from spacex_falcon.evaluation import error_analysis, feature_importance, logistic_coefficients, run_experiments

data = modeling_frame(load_launch_data())
results = run_experiments(data)
```

## Calibration

The raw and sigmoid calibrated logistic models are compared using Brier score and log loss. Calibration is retained only if it improves out of period probability quality.

```python
results.calibration[['Model', 'ROCAUC', 'Brier', 'LogLoss']]
```
